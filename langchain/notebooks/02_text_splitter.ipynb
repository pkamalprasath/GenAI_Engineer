### How to recursively split text by character 
### This is recommended one for the generic text. It is parameterized by a list of characters. It tries to split on them in order untill the chunks are small enough. 
### how the text is split by list of characters - "\n\n,\n, "", " " 
### how the chunk size is measured : by numbers of characters .

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter 
text_splitters=RecursiveCharacterTextSplitter(chunk_size=500,chunk_overlap=50)
final_documents=text_splitters.split_documents(pdf)
print(final_documents)
```

```python
stories=""
with open("stories.txt") as f:
    stories=f.read()

text_splitters=RecursiveCharacterTextSplitter(chunk_size=100,chunk_overlap=10)
text=text_splitters.create_documents([stories])
print(text[0])
```

```python
### How to character - character text split 
### This is .simple method. the splits based on a given character sequence,which defaults "\n\n". Chunk is measured by number of character 
### how the text by single text spliter "\n\n" 
### how the chunk size is measured : by numbers of characters .
```

```python
# Load the text file        
from langchain_community.document_loaders import TextLoader
doc=TextLoader('stories.txt').load() 
print(doc)
```

```python
from langchain_text_splitters import CharacterTextSplitter 
text_splitters=CharacterTextSplitter(separator="\n\n",chunk_size=100,chunk_overlap=10)
text=text_splitters.split_documents(doc)
print(text[0])
print(text[1])
```

What is HTML Text Splitting?

HTML text splitting is the process of:

extracting useful readable content from an HTML page

removing tags, scripts, navigation, ads

splitting the content into smaller, meaningful chunks

```python
from langchain_text_splitters import HTMLHeaderTextSplitter

html_string="""
<!DOCTYPE html>
<html lang="en">
<head>
    <meta charset="UTF-8">
    <title>Understanding Attention Mechanisms</title>
    <style>
        body { font-family: Arial; }
        nav { background: #eee; padding: 10px; }
    </style>
</head>

<body>

<nav>
    <a href="#">Home</a> |
    <a href="#">Articles</a> |
    <a href="#">Contact</a>
</nav>

<header>
    <h1>Attention Mechanisms in Deep Learning</h1>
    <p><em>Published on July 10, 2025</em></p>
</header>

<section>
    <h2>Introduction</h2>
    <p>
        Attention mechanisms allow neural networks to focus on the most relevant
        parts of the input while processing data. They are widely used in
        natural language processing and computer vision.
    </p>
</section>

<section>
    <h2>Why Attention is Important</h2>
    <p>
        Traditional models process inputs sequentially, which makes it hard
        to capture long-range dependencies. Attention solves this by
        directly relating different parts of the input.
    </p>
    <ul>
        <li>Improves model accuracy</li>
        <li>Handles long sequences efficiently</li>
        <li>Enables parallel computation</li>
    </ul>
</section>

<section>
    <h2>Types of Attention</h2>

    <h3>Self-Attention</h3>
    <p>
        Self-attention allows a sequence to attend to itself. It is a core
        component of Transformer models such as BERT and GPT.
    </p>

    <h3>Causal Attention</h3>
    <p>
        Causal attention ensures that future tokens are masked, making it
        suitable for text generation tasks.
    </p>
</section>

<section>
    <h2>Transformers and Attention</h2>
    <p>
        Transformers rely entirely on attention mechanisms, removing the need
        for recurrent or convolutional layers.
    </p>
</section>

<footer>
    <p>© 2025 AI Learning Blog</p>
</footer>

<script>
    console.log("This script should be ignored by the text splitter");
</script>

</body>
</html>
""" 

headers_to_split_on=[
    ("h1","Header 1"),
    ("h2","Header 2"),
    ("h3","Header 3")
]

html_split=HTMLHeaderTextSplitter(headers_to_split_on)
html=html_split.split_text(html_string)
html
```

```python
url = "https://plato.stanford.edu/entries/goedel"

headers_to_split_on=[
    ("h1","Header 1"),
    ("h2","Header 2"),
    ("h3","Header 3"),
    ("h4","Header 4"),
]
html_split=HTMLHeaderTextSplitter(headers_to_split_on)
html=html_split.split_text_from_url(url)
html
```

```python
""" What is a JSON Text Splitter?

A JSON text splitter is used to break large JSON data into smaller, meaningful chunks while preserving structure and keys.
Instead of treating JSON as plain text, it respects nested fields (objects, lists) so chunks remain semantically useful.

What is a JSON Text Splitter?

A JSON text splitter is used to break large JSON data into smaller, meaningful chunks while preserving structure and keys.
Instead of treating JSON as plain text, it respects nested fields (objects, lists) so chunks remain semantically useful. 

When to Use JSON Splitter
Use it when:
Working with API responses
Storing structured documents
Building RAG systems with metadata
Ingesting logs, configs, knowledge bases """ 
```

```python
import json 
import requests 
json_data=requests.get("https://api.smith.langchain.com/openapi.json").json()
```

```python
json_data
```

```python
from langchain_text_splitters import RecursiveJsonSplitter
json_spiltter=RecursiveJsonSplitter(max_chunk_size=300)
json=json_spiltter.split_json(json_data)
print(json[0])
```

```python
for chunks in json[:3]:
    print(chunks)
```

```python
## the splitter can also output documents
docs=json_spiltter.create_documents(texts=[json_data])
for doc in docs[:3]:
    print(doc)
```

```python

texts=json_spiltter.split_text(json_data)
print(texts[0])
print(texts[1])
```
