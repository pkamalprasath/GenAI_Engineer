```python
from langchain_chroma import Chroma
from langchain_community.document_loaders import TextLoader
from langchain_text_splitters import RecursiveCharacterTextSplitter
from langchain_community.embeddings import OllamaEmbeddings

data=TextLoader("stories.txt").load()
text_splitter=RecursiveCharacterTextSplitter(chunk_size=500,chunk_overlap=50)
splits=text_splitter.split_documents(data)
embeddings=OllamaEmbeddings(model="qwen3-embedding:0.6b")
vectordb=Chroma.from_documents(documents=splits,embedding=embeddings)

```

Quering the chroma db

```python
input_query = "Slowly, Arun began to improve. He fixed small bugs"
inference = vectordb.similarity_search(input_query)
inference[0].page_content
```

Save the chromadb

```python
vectordb=Chroma.from_documents(documents=splits,embedding=embeddings,persist_directory="./chroma_db")
```

Load the data from the file that is saved


```python
db2=Chroma(persist_directory="./chroma_db",embedding_function=embeddings)
```

Search in the stores file

```python
inference = db2.similarity_search(input_query)
inference[0].page_content
```

Retriever option

```python
retriver=vectordb.as_retriever()
retriver.invoke(input_query)[0].page_content
```
