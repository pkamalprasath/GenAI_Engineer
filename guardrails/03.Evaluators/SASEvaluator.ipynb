```python
### *Short Answer Similarity (SAS) Evaluator* 
 
What it is The SAS (Short Answer Similarity) Evaluator measures how semantically similar a model’s predicted answer is to a ground-truth answer.
Instead of exact string matching, it uses sentence embeddings (e.g., SentenceTransformers) to compare meaning.

### Pros (Why SAS is useful)

- Paraphrase-friendly
- Rewards answers that are meaning-equivalent even if wording is different.
- Better than exact match
- Avoids penalizing correct answers just because phrasing differs.
- Model-agnostic
- Works with any generator (OpenAI, LLaMA, etc.).
- Fast & cheap
- No LLM calls — just embedding similarity.
- Great for QA benchmarks
- Especially useful for datasets like SQuAD, PubMedQA, Natural Questions.

### Cons (Where SAS can mislead)

- Does NOT check factual grounding
- A hallucinated answer can score high if it sounds similar to the ground truth.
- Insensitive to evidence source
- It doesn’t care where the answer came from — only how similar it sounds.
- Weak on numeric or precise facts
- Small numerical errors may still score high.
- Embedding-model dependent
- Different sentence-transformer models → different scores.
```

```python
from haystack.components.evaluators import SASEvaluator
sas_evaluator = SASEvaluator()
sas_evaluator.warm_up()
result = sas_evaluator.run(ground_truth_answers=["Berlin", "Paris"], predicted_answers=["Berlin", "Lyon"])
print(result["individual_scores"])
print(result["score"])
```
