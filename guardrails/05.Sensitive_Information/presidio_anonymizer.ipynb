### Analyzer with Pattern recognizer with entity parameter

```python
from presidio_analyzer import AnalyzerEngine, PatternRecognizer

# Text to analyze
text_to_anonymize = "His name is Mr. Jones and his phone number is 212-555-5555"

# Create analyzer engine
analyzer = AnalyzerEngine()

# Run analysis
# IMPORTANT: entities must be a list, not a single string
results = analyzer.analyze(
    text=text_to_anonymize,
    entities=["PHONE_NUMBER"],
    language="en"
)

print(results)
```

```python
from presidio_anonymizer import AnonymizerEngine 
from presidio_anonymizer.entities import OperatorConfig 
```

### 1. Masking

```python
anonymizer=AnonymizerEngine()
anonymized_results=anonymizer.anonymize(text=text_to_anonymize,analyzer_results=results,
                                        operators={
                                            "PHONE_NUMBER":OperatorConfig(
                                                "mask",{
                                                "type"         : "mask",
                                                "masking_char" : "*",
                                                "chars_to_mask": 12,
                                                "from_end"     : True
                                                })})
print(anonymized_results.text)                                      
```

### 2. Redaction

```python
anonymizer=AnonymizerEngine()
anonymized_results=anonymizer.anonymize(text=text_to_anonymize,analyzer_results=results,
                                        operators = {"PHONE_NUMBER": OperatorConfig("redact")
})
print(anonymized_results.text)                                      
```

### 3.Replace with fixed value

```python
anonymizer=AnonymizerEngine()
anonymized_results=anonymizer.anonymize(text=text_to_anonymize,analyzer_results=results,
                                        operators = {"PHONE_NUMBER": 
                                        OperatorConfig("replace",{"new_value": "[PHONE_REMOVED]"})})
print(anonymized_results.text)                                      
```

### 4.Hashing (irreversible, secure)

```python
anonymizer=AnonymizerEngine()
anonymized_results=anonymizer.anonymize(text=text_to_anonymize,analyzer_results=results,
                                        operators = {"PHONE_NUMBER": OperatorConfig("hash")
})
print(anonymized_results.text)                                      
```

### Pseudonymization (fake but realistic)

```python
anonymizer=AnonymizerEngine()
anonymized_results=anonymizer.anonymize(text=text_to_anonymize,analyzer_results=results,
                                            operators = {"PHONE_NUMBER": OperatorConfig(
                                                    "replace", {"new_value": "999-999-9999"})})
print(anonymized_results.text)                    
```
