```python
# download_edufineweb.py
import os
from huggingface_hub import snapshot_download
# ← replace with your actual HF repo
REPO_ID   = "VizuaraAI/edufineweb"
HF_TOKEN  = "YOUR_HF_TOKEN"
# ← this is the exact folder you already have
DEST_DIR  = "/workspace/build-nanogpt/edu_fineweb10B"
os.makedirs(DEST_DIR, exist_ok=True)
print(f"Downloading all .npy files from {REPO_ID} into {DEST_DIR} …")
snapshot_download(
    repo_id    = REPO_ID,
    repo_type  = "dataset",
    local_dir  = DEST_DIR,
    token      = HF_TOKEN,
    allow_patterns = ["*.npy"],    # only grab your .npy files
)
print("✅ All files saved to", DEST_DIR)
```
