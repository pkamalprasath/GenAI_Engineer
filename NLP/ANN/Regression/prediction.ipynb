```python
#importing all the important packages, lib for this implementation
import pandas as pd 
import numpy as np
from tensorflow.keras.models import load_model
import pickle
```

```python
### load the trained model, scalar pickel file, one hot encoder.
model=load_model('model.h5')

with open('label_encoder_gender.pk1','rb') as file:
    label_encoder_gender=pickle.load(file)

with open('onehot_encoder_geo.pk1','rb') as file:
    onehot_encoder_geo=pickle.load(file) 

with open('scaler.pk1','rb') as file:
    scaler=pickle.load(file) 
```

```python
# Example input data
input_data = {
    'CreditScore': 600,
    'Geography': 'France',
    'Gender': 'Male',
    'Age': 40,
    'Tenure': 3,
    'Balance': 60000,
    'NumOfProducts': 2,
    'HasCrCard': 1,
    'IsActiveMember': 1,
    'EstimatedSalary': 50000
}
```

```python
input_df=pd.DataFrame([input_data])
input_df
```

```python
# One-hot encode 'Geography'
geo_encoded = onehot_encoder_geo.transform([[input_data['Geography']]]).toarray()
geo_encoded_df = pd.DataFrame(geo_encoded, columns=onehot_encoder_geo.get_feature_names_out(['Geography']))
```

```python
## Encode categorical variables
input_df['Gender']=label_encoder_gender.fit_transform(input_df['Gender'])
## concatination one hot encoded 
input_df=pd.concat([input_df.drop("Geography",axis=1),geo_encoded_df],axis=1)
## Scaling the input data
input_scaled=scaler.transform(input_df)
input_scaled
```

```python
prediction=model.predict(input_scaled)
prediction
```

```python
prediction_proba = prediction[0][0]
print(prediction_proba) 
if prediction_proba > 0.5:
    print('The customer is likely to churn.')
else:
    print('The customer is not likely to churn.')
```
