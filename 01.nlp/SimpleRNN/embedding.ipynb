# Embedding in RNN (Simple Explanation)
# ------------------------------------
# Embeddings are a way to convert words (or categories) into numeric vectors
# so that the RNN can understand them.
#
# Instead of representing each word as a one-hot vector (very long, mostly zeros),
# an embedding layer maps each word to a dense, small vector like [0.12, -0.5, 0.9].
#
# Why embeddings?
# 1. They capture meaning — similar words get similar vectors.
# 2. They reduce input size — faster and easier for the RNN to learn.
# 3. They help the RNN understand patterns in text.
#
# Example:
# "king" and "queen" will have vectors that are close to each other.
#
# In simple terms:
# Embedding = "word meaning stored as numbers"
#
# RNNs use these embeddings as input at each time step instead of raw text.

```python
from tensorflow.keras.preprocessing.text import one_hot
```

```python
### sentences
sent=[  'the glass of milk',
     'the glass of juice',
     'the cup of tea',
    'I am a good boy',
     'I am a good developer',
     'understand the meaning of words',
     'your videos are good',]
```

```python
## Define the vocabulary size
voc_size=10000
```

```python
### One Hot Representation
one_hot_retr=[one_hot(words,voc_size) for words in sent] 
```

```python
one_hot_retr
```

# PCA (Principal Component Analysis)
# ----------------------------------
# PCA reduces many features into fewer, important ones.
# It finds the directions where data varies the most and keeps only those.
#
# Why use it?
# - Removes noise
# - Makes data smaller and faster for models
#
# Simple meaning:
# PCA = "smart feature compression" before sending data into the RNN.

```python
## word Embedding Representation

# Embedding layer converts word indices into dense vectors.
# It turns each word into a meaningful numeric vector that the RNN can learn from.
from tensorflow.keras.layers import Embedding

# pad_sequences makes all sentences the same length.
# RNNs need fixed-length sequences, so padding adds zeros to shorter sentences.
from tensorflow.keras.utils import pad_sequences

# Sequential lets us build the model layer-by-layer.
# Useful for simple models like Embedding → LSTM → Dense.
from tensorflow.keras.models import Sequential
import numpy as np 
```

```python
# sent_length = 8 → final size of each sequence
# pad_sequences → makes all sequences same length (pads with zeros)
# padding='pre' → zeros added at the beginning
# Result: embedded_docs = uniform-length input for Embedding/RNN

sent_length=8
embedded_docs=pad_sequences(one_hot_retr,padding='pre',maxlen=sent_length)
print(embedded_docs)
```

```python
# Feature representation - Input 2 
dim=10

model=Sequential()
model.add(Embedding(voc_size,dim,input_length=sent_length))
model.compile('adam','mse')

# Embedding layer explanation (short)
# -----------------------------------
# dim = 10 → size of each word’s embedding vector.
# Embedding(voc_size, dim) → creates a lookup table where each word ID
# is mapped to a learnable 10-dimensional vector.
# input_length=sent_length → ensures RNN gets fixed-length input.

# Model not built yet because it needs actual input data first.
# Once model.fit() or model.predict() is called, the embedding layer will build
# and param count = voc_size * dim.
```
