# ContextRelevanceEvaluator

## What it is
The **ContextRelevanceEvaluator** measures **how relevant the retrieved context documents are to a given question**, without considering the generated answer. 
It evaluates **retriever quality**, not language-model generation.

---
## Why it exists
In Retrieval-Augmented Generation (RAG) systems, many failures happen **before** generation:
- the retriever fetches irrelevant documents
- the LLM is forced to answer using weak or noisy context
This evaluator answers:
> **“Did my retriever fetch documents that are actually useful for this question?”**
---
## How it works (high-level)
- Takes a **question** and a set of **retrieved context documents**
- Computes **semantic similarity** between the question and the contexts (embedding-based)
- Produces a **relevance score** per query (and aggregated metrics)

>  No generated answer is required.

---
## Pros (Strengths)
- **Direct retriever evaluation**  
  Measures retrieval quality independently of the LLM.

- **Answer-agnostic**  
  Works even when generation is disabled or incorrect.

- **Fast and inexpensive**  
  Uses embeddings instead of LLM calls.

- **Excellent for RAG debugging**  
  Helps tune:
  - embedding models
  - chunk sizes
  - indexing strategy
  - `top_k` values

---
##  Cons (Limitations)

- **Does not evaluate answer correctness**  
  Relevant context does not guarantee a correct answer.

- **Semantic, not factual**  
  Documents may be topically related but miss key facts.

- **Embedding-model dependent**  
  Scores vary with different sentence-transformer models.

- **Cannot detect hallucinations**  
  High relevance does not mean the answer is grounded.

---

## Best Practice in RAG Evaluation

Use **ContextRelevanceEvaluator** together with:

| Evaluator | What it Measures |
|---------|------------------|
| ContextRelevance | Retriever quality |
| Faithfulness | Grounding to retrieved context |
| SAS (Short Answer Similarity) | Answer similarity to ground truth |

Together, they provide **end-to-end RAG diagnostics**.
---

## When to Use

- Tuning retrieval pipelines  
- Comparing embedding models  
- Diagnosing poor RAG answers  
- Separating retriever issues from gene


```python
from haystack.components.evaluators import ContextRelevanceEvaluator

questions = ["Who created the Python Programming Language?"]

contexts = [
    [
        "Python, created by Guido van Rossum in the late 1980s, is a high-level general-purpose programming language. "
        "Its design philosophy emphasizes code readability, and its language constructs aim to help programmers write clear, logical code for both small and large-scale software projects."
    ],
]

evaluator = ContextRelevanceEvaluator()
result = evaluator.run(questions=questions, contexts=contexts)
print(result["score"])
print(result["individual_scores"])
print(result["results"])
```
